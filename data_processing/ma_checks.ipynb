```python
# Import packages
from dataclasses import dataclass
import numpy as np
import os
import pandas as pd

# Linting
%load_ext pycodestyle_magic
%pycodestyle_on

# Set the maximum number of columns to 100
pd.set_option('display.max_columns', 100)
pd.set_option('display.max_rows', 100)
```

```python
# Set paths and filenames
@dataclass(frozen=True)
class Paths:
    '''Singleton object for storing paths to data and database.'''

    data_path: str = './data'
    data_filename: str = 'SAMueL ssnap extract v2.csv'
    data_save_path: str = './output'
    data_save_filename: str = 'dummy.csv'
    issues_filename: str = 'cleaning_issues_log.csv'


paths = Paths()
```

```python
# Load data
raw_data: pd.DataFrame = pd.read_csv(
    os.path.join(paths.data_path, paths.data_filename), low_memory=False)
```

```python
summary = pd.DataFrame(raw_data.dtypes, columns=['dtype'])
# Get total patients
summary['Total patients'] = len(raw_data)
# Get number of missing values for each column
summary['Missing values'] = raw_data.isnull().sum()
# Get percentage of missing values for each column
summary['Missing percentage'] = raw_data.isnull().mean() * 100
summary['Complete percentage'] = 100 - summary['Missing percentage']
# Round percentages to 3 decimal places
summary['Missing percentage'] = summary['Missing percentage'].round(3)
summary['Complete percentage'] = summary['Complete percentage'].round(3)
# Remove dtype
summary = summary.drop(columns=['dtype'])
# Save summary to csv
summary.to_csv(os.path.join(paths.data_save_path, 'data_summary.csv'))
# Save to LaTeX
with open(os.path.join(paths.data_save_path, 'data_summary.tex'), 'w') as f:
    f.write(summary.to_latex())
```

```python
# Load cleaning issues log
issues_log = pd.read_csv(os.path.join(paths.data_save_path, paths.issues_filename))
```

```python
issue_summary = issues_log.describe()
# Save issue summary to csv
issue_summary.to_csv(os.path.join(paths.data_save_path, 'issue_summary.csv'))
```
