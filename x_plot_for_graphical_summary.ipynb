```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 360, 500)
log_odds = 0.91 - (x * 0.91 / 356)      # linear decay in log-odds space
odds = np.exp(log_odds)           # convert to odds ratio


if False:
    with plt.xkcd():
        fig, ax = plt.subplots(figsize=(6, 5))
        ax.plot(x, odds, linewidth=2)
        ax.axhline(1, color='gray', linewidth=1, linestyle='--', label='Odds ratio = 1 (no effect)')
        ax.axvline(0, color='black', linewidth=1)
        ax.set_xlabel('Time (minutes)')
        ax.set_ylabel('Benefit of thrombolysis\n(odds ratio in achieving mRS 0-1)')
        ax.set_title('Decay of Thrombolysis Effectiveness with Time', pad=20)
        ax.set_xlim(0, 360)
        ax.legend(fontsize=8)
        plt.tight_layout()
        plt.savefig('xkcd_plot.png', dpi=300)
        plt.show()
else:
    fig, ax = plt.subplots(figsize=(5, 4))
    ax.plot(x, odds, linewidth=2)
    ax.axhline(1, color='gray', linewidth=1, linestyle='--', label='Odds ratio = 1 (no effect)')
    ax.axvline(0, color='black', linewidth=1)
    ax.set_xlabel('Time from onset (minutes)')
    ax.set_ylabel('Benefit of thrombolysis\n(odds ratio in achieving mRS 0-1)')
    #ax.set_title('Decay of Thrombolysis Effectiveness with Time', pad=20)
    ax.set_xlim(0, 360)
    ax.legend(fontsize=9)
    plt.tight_layout()
    plt.savefig('plot.png', dpi=300)
    plt.show()

    
```
